```python
# Força backend Win32
try:
    app = Application(backend="win32").connect(title_re=".*Módulos.*")
    print("✓ Conectado com Win32")

    window = app.window(title_re=".*Módulos.*")
    print("\n" + "="*60)
    print("ESTRUTURA COM WIN32:")
    print("="*60)
    window.print_control_identifiers()

except Exception as e:
    print(f"❌ Erro: {e}")
```
